```python
x = np.arange(0, 4, 0.05)
y = np.sin(x  * np.pi)

plt.figure(figsize = (8,5))
plt.plot(x,y)
plt.xlabel("Time [s]")
plt.ylabel("Amplitude [V]")
plt.tilte("Sine wave")
plt.xlabel(x)
plt.ylabel("f(x)")
plt.xlim(-1.5,1.4)
plt.ylim(-0.5, 0.5)
plt.legend()
plt.grid(True)
plt.show()
```
